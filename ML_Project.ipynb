```python
# Check Outlier
plt.figure(figsize=(8,6))
sns.boxplot(x="age", data=df, color="teal")
plt.title("Boxplot of age")
plt.show()
```

```python
# Outlier Removal
# Find q1, q3 and IQR

Q1 = df["age"].quantile(0.25)
Q3 = df["age"].quantile(0.75)
IQR = Q3 - Q1
# Calculate upper and lower range
lower_range = Q1 - 1.5 * IQR
upper_range = Q3 + 1.5 * IQR
print(lower_range)
print(upper_range)
```

```python
plt.figure(figsize=(8,6))
sns.boxplot(x = "age", data = df, color = "teal")
plt.title("Boxplot of age")
plt.show()
```

```python
df["age"] = np.where(df["age"] > upper_range, upper_range, df["age"])
df["age"] = np.where(df["age"] < lower_range, lower_range, df["age"])
df
```

```python
plt.figure(figsize=(8,6))
sns.boxplot(x = "age", data = df, color = "teal")
plt.title("Boxplot of age")
plt.show()
```

```python
# Outlier Removal
# Find q1, q3 and IQR

Q1 = df["fare"].quantile(0.25)
Q3 = df["fare"].quantile(0.75)
IQR = Q3 - Q1
# Calculate upper and lower range
lower_range = Q1 - 1.5 * IQR
upper_range = Q3 + 1.5 * IQR
print(lower_range)
print(upper_range)
```

```python
plt.figure(figsize=(8,6))
sns.boxplot(x = "fare", data = df, color = "teal")
plt.title("Boxplot of fare")
plt.show()
```

```python
from sklearn.model_selection import train_test_split
x = df.drop("survived", axis=1)
y = df["survived"]
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=None)

print(x)
print(y)
```

```python
print(x_train.shape)
print(x_test.shape)
print(y_train.shape)
print(y_test.shape)
```

```python
from sklearn.tree import DecisionTreeClassifier

# Initialize the Decision Tree Classifier
# max_depth controls overfitting; criterion can be "gini" or "entropy"
clf = DecisionTreeClassifier()   #(max_depth=3, criterion='gini', ranom_sate = 42)

# Convert y_train and y_test to integer type as DecisionTreeClassifier expects discrete classes
y_train = y_train.astype(int)
y_test = y_test.astype(int)

# Convert categorical columns in x_train and x_test to numerical using LabelEncoder
from sklearn.preprocessing import LabelEncoder

# Identify non-numeric columns in x_train
non_numeric_cols_train = x_train.select_dtypes(include=['category', 'bool']).columns
for col in non_numeric_cols_train:
    le = LabelEncoder()
    x_train[col] = le.fit_transform(x_train[col])
    # Ensure x_test columns are transformed using the same encoder fitted on x_train
    x_test[col] = le.transform(x_test[col])

# Train the classifier on the training data
clf.fit(x_train, y_train)

# Make predictions on the test data
y_pred = clf.predict(x_test)
```
